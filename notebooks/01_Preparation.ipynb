#### Step 2. Filter templates

Now, we are filtering the grid model templates according to their position related to the main sequence and the effective temperature on the model. We can define the parameters and call the function directly. 

```python
import os
import shutil
import zipfile

os.chdir("../functions")
import filt
```

The function `filter_templates` select only the templates above MS and with temperatures above a limit user fix. Then it creates a new directory in the same path where save these new filtered model grid. We recomend to zip from `functions` folder original grids and substitute by these new ones. 

```python
# Parameters of the function:

listInstruments= ['GAIA','2MASS','WISE']  # instruments of using filters
directories= ['s---s-i','sp--s-i','sp--h-i','s---smi','sp--smi','sp--hmi',
              's-p-smi','s-p-hmi','s-pbsmi','s-pbhmi','s-u-smi','s-u-hmi',
              's-ubsmi','s-ubhmi','spu-smi','spu-hmi','spubsmi','spubhmi']       # where our family model sets are
tempe_lim= 2000 # K limit of temperature for our models. According to our study, 2000 K is the optimal limit
outDir= './aboveMS'  # where to save the new model grid

filt.filter_templates.filter_aboveMS(listInstruments, directories, tempe_lim, outDir)
```

Optionally, we can implement this function to change family folders in our directory from originals to this new ones. Original grid will be zip in a new file `original_grid.zip`

```python
original_zip= './original_grid.zip'

filt.change_grid(original_zip, outDir)
```
